## 1. Export

I select only the columns needed for the dashboard from the SQLite database and save them as a small CSV file.

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('../data/home_credit.db')

query = '''
SELECT
    TARGET,
    AGE_YEARS,
    AGE_GROUP,
    NAME_INCOME_TYPE,
    NAME_EDUCATION_TYPE,
    NAME_CONTRACT_TYPE,
    CODE_GENDER,
    AMT_INCOME_TOTAL,
    AMT_CREDIT,
    DAYS_EMPLOYED_MISSING
FROM applications
'''

df_tableau = pd.read_sql(query, conn)
df_tableau.to_csv('../outputs/tableau_data.csv', index=False)
df_tableau.shape
```
